```python
# Para agregar datos mediante transformación basada en una función:

# 1. Definimos la función:
import random
def calcularGanancias(retweets):
    ganancia = retweets * random.randint(3,5)
    return ganancia

# 2. Agregamos la columna calculada:
df["ganancias"] = df["retweets"].apply(calcularGanancias)

df.head(10)
```

```python
# Para aplicar una transformación basada en más de una columna:

def popularidad(fila):
    resultado = fila["followees"]/fila["followers"]
    return resultado

df["popularidad"] = df.apply(popularidad,axis=1)

df.head()
```

```python
# Para aplicar funciones de agregación en varias columnas agrupando por una columna hacemos lo siguiente:

grouped = df.groupby("country").agg({
    "followers": 'sum',
    "mentions": 'mean',
    "retweets": 'max'
})

grouped
# Nota: La columna por la que se referencía se convierte en el nuevo identificador del dataframe obtenido
```

```python
# Para graficar por medio de matplotlib:

import matplotlib.pyplot as plt

grouped["followers"].plot()

plt.show()
```

```python
# Para graficar en barras:

grouped["followers"].plot(kind="bar")

plt.show()
```
